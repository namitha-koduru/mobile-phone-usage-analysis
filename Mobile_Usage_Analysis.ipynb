```python
import pandas as pd

df = pd.read_csv("user_behavior_dataset-selected-columns.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
# Check duplicates
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Average App Usage Time:")
print(df["App Usage Time (min/day)"].mean())

print("\nAverage Screen On Time:")
print(df["Screen On Time (hours/day)"].mean())

print("\nAverage Battery Drain:")
print(df["Battery Drain (mAh/day)"].mean())
```

```python
import matplotlib.pyplot as plt

df["Operating System"].value_counts().plot(kind="bar")

plt.title("Operating System Distribution")
plt.xlabel("Operating System")
plt.ylabel("Number of Users")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Screen On Time (hours/day)"])

plt.title("Screen On Time Distribution")
plt.xlabel("Hours per Day")
plt.ylabel("Users")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Screen On Time (hours/day)"],
    df["Battery Drain (mAh/day)"]
)

plt.title("Screen Time vs Battery Drain")
plt.xlabel("Screen Time (hours/day)")
plt.ylabel("Battery Drain (mAh/day)")

plt.show()
```

```python
from google.colab import files

files.download("os_distribution.png")
files.download("screen_time_distribution.png")
files.download("battery_vs_screen_time.png")
```

```python
plt.figure(figsize=(6,6))

df["Operating System"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Operating System Share")
plt.ylabel("")

plt.savefig("os_share_pie.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(6,6))

df["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")
plt.ylabel("")

plt.savefig("gender_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))

sns.boxplot(y=df["Age"])

plt.title("Age Distribution and Outliers")

plt.savefig("age_boxplot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.violinplot(
    x="Operating System",
    y="Screen On Time (hours/day)",
    data=df
)

plt.title("Screen Time Distribution by OS")

plt.savefig("screen_time_violin.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Age"],
    df["App Usage Time (min/day)"]
)

plt.title("Age vs App Usage")
plt.xlabel("Age")
plt.ylabel("App Usage Time")

plt.savefig("age_vs_app_usage.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(10,5))

df["Device Model"].value_counts().plot(kind="bar")

plt.title("Device Model Distribution")
plt.xlabel("Device Model")
plt.ylabel("Users")

plt.savefig("device_model_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(10,6))

corr = df.select_dtypes(include=["int64","float64"]).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
sns.pairplot(
    df[
        [
            "App Usage Time (min/day)",
            "Screen On Time (hours/day)",
            "Battery Drain (mAh/day)",
            "Data Usage (MB/day)"
        ]
    ]
)

plt.savefig("pairplot.png")
plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.title("Missing Values Heatmap")

plt.savefig("missing_values_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```
